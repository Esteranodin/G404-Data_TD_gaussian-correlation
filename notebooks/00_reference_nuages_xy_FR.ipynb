```python
import pandas as pd
import plotly.express as px

csv_path = "../lab/gaussian_project_solution/data/purchase_transactions.csv"
transactions = pd.read_csv(csv_path)

print(transactions.shape)
transactions.head()
```

```python
plot_specs = {
    "checkout_wait_minutes": "Attente en caisse (minutes)",
    "distance_to_store_km": "Distance domicile–magasin (km)",
    "age": "Âge (ans)",
}
shared_y_range = [
    transactions["purchase_amount"].min(),
    transactions["purchase_amount"].max(),
]

# Un graphique par variable X, avec les mêmes repères visuels.
for x_column, x_label in plot_specs.items():
    figure = px.scatter(
        transactions,
        x=x_column,
        y="purchase_amount",
        color="group",
        opacity=0.4,
        range_y=shared_y_range,
        labels={
            x_column: x_label,
            "purchase_amount": "Montant d'achat (€)",
            "group": "Magasin",
        },
        color_discrete_map={"A": "#0072B2", "B": "#D55E00"},
        title=f"{x_label} et montant d'achat",
    )
    figure.show()
```

## À décrire

Pour chaque nuage, décrivez d'abord A+B, puis A et B séparément :
tendance, dispersion, chevauchement et position des groupes.
Le graphique seul ne permet pas d'affirmer un mécanisme ou une
causalité.
